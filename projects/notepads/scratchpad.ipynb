Two complement system:

```python
def to_two_complement(n, bits, positive = True):
    if positive:
        res = n
    else:
        res = 2 ** bits - n
    return res

def stringify(n, bits):
    c = bits
    res = ''
    while c > 0:
        b = n & 1
        n = n >> 1
        c -= 1
        res += str(b)
    res = res[::-1]
    if n > 0:
        res = res +'(o)'
    return res

def printout_two_complement(n, bits):
    negComp = to_two_complement(n, bits, False)
    print(f'n: {n}')
    print(stringify(n, bits))
    print(stringify(negComp, bits))
    print(stringify(n + negComp, bits))




#for i in range(1, 9):
#    printout_two_complement(i, 4)

printout_two_complement(7, 16)


```
